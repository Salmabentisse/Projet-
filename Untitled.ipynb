```python
from tensorflow.keras.utils import to_categorical

X_train_dl = X_train / 255.0
X_test_dl = X_test / 255.0

y_train_dl = to_categorical(y_train, 10)
y_test_dl = to_categorical(y_test, 10)
```

```python
X_train_dl = X_train_dl.reshape(-1, 28*28)
X_test_dl = X_test_dl.reshape(-1, 28*28)
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense

model = Sequential([
    Dense(128, activation='relu', input_shape=(784,)),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])
```

```python
model.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)
```

```python
history = model.fit(
    X_train_dl,
    y_train_dl,
    validation_split=0.2,
    epochs=5,
    batch_size=32
)
```

```python
loss, acc = model.evaluate(X_test_dl, y_test_dl)
print("Test Accuracy:", acc)
```

```python
import matplotlib.pyplot as plt

plt.plot(history.history['accuracy'], label='train')
plt.plot(history.history['val_accuracy'], label='val')
plt.title("Accuracy du modèle")
plt.legend()
plt.show()
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense, Dropout

# Modèle amélioré
model_opt = Sequential([
    Dense(256, activation='relu', input_shape=(784,)),
    Dropout(0.3),
    Dense(128, activation='relu'),
    Dropout(0.3),
    Dense(64, activation='relu'),
    Dense(10, activation='softmax')
])

# Compilation
model_opt.compile(
    optimizer='adam',
    loss='categorical_crossentropy',
    metrics=['accuracy']
)

# Entraînement
history_opt = model_opt.fit(
    X_train_dl,
    y_train_dl,
    validation_split=0.2,
    epochs=10,
    batch_size=32
)
```

```python
import matplotlib.pyplot as plt

# Évaluation
loss, acc = model_opt.evaluate(X_test_dl, y_test_dl)
print("Optimized Model Accuracy:", acc)

# Comparaison des modèles
print("Ancien modèle:", history.history['val_accuracy'][-1])
print("Optimisé:", history_opt.history['val_accuracy'][-1])

# Courbes
plt.plot(history.history['val_accuracy'], label='Simple MLP')
plt.plot(history_opt.history['val_accuracy'], label='Optimized MLP')
plt.title("Comparaison des modèles")
plt.legend()
plt.show()
```

```python
import numpy as np
from sklearn.metrics import confusion_matrix
import seaborn as sns
import matplotlib.pyplot as plt

# Prédictions
y_pred_dl = model_opt.predict(X_test_dl)
y_pred_classes = np.argmax(y_pred_dl, axis=1)

y_true = np.argmax(y_test_dl, axis=1)

# Matrice de confusion
cm = confusion_matrix(y_true, y_pred_classes)

plt.figure(figsize=(8,6))
sns.heatmap(cm, cmap="Blues")
plt.title("Matrice de confusion - Deep Learning")
plt.xlabel("Prédit")
plt.ylabel("Vrai")
plt.show()
```

```python
# Accuracy finale
test_loss, test_acc = model_opt.evaluate(X_test_dl, y_test_dl)
print("Final Test Accuracy:", test_acc)

# Afficher quelques erreurs
errors = np.where(y_pred_classes != y_true)[0]

plt.figure(figsize=(10,5))

for i, idx in enumerate(errors[:10]):
    plt.subplot(2,5,i+1)
    plt.imshow(X_test[idx], cmap="gray")
    plt.title(f"V:{y_true[idx]} P:{y_pred_classes[idx]}")
    plt.axis("off")

plt.show()
```
